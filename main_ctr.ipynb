# CTR (대조군) optim — FC_AAL_ComBat_all_163.mat

PD 파이프라인(`main_ppmi_pd.py`)을 그대로 재사용하고 **데이터 소스만 CTR로 교체**한다.

- 입력: `data/AALv3/FC_AAL_ComBat_all_163.mat` 의 `group=='CTR'` (57명)
- **CTR idx 0 = subject 100890**
- 결과 저장: `output_ppmi_ctr/<sub_num>/` (PD 결과와 분리)
- 설정: PD와 동일 (activity=1.0, freeze=False, rww_ 배선, FIC→EIB→Part3)

그냥 아래 셀 실행하면 CTR idx 0 optim 시작. 다른 CTR subject는 `CTR_IDX` 변경(0~56).

```python
import sys
import main_ppmi_pd as M

# ===== CTR 데이터 소스로 override (PD 파이프라인 코드는 그대로) =====
M.MAT_PATH     = "data/AALv3/FC_AAL_ComBat_all_163.mat"   # 전체(PD+CTR) 파일
M.GROUP_FILTER = "CTR"                                     # 대조군 57명만 필터
M.OUTPUT_DIR   = "output_ppmi_ctr"                         # CTR 결과 분리 저장

# cache_version 라벨과 CTR 명시(경로는 sub_num으로 이미 분리되므로 안전장치)
_orig_make_config = M.make_config
def _ctr_make_config(p, idx):
    cfg = _orig_make_config(p, idx)
    cfg.cache_version = f"v_ctraal163_tr25_s{idx}"
    return cfg
M.make_config = _ctr_make_config

# ===== CTR idx 0 optim 실행 =====
CTR_IDX = 0   # 0~56 (CTR idx0 = sub 100890)
sys.argv = ["main_ppmi_pd", "--subject-idx", str(CTR_IDX)]
M.main()
```
