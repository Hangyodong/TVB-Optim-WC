# DBS sweep — idx 4, 5, 6

`dbs_from_cache.py` 재사용: **Part3 grad 캐시**를 로드해 DBS 자극 sweep (FIC/EIB/Part3 재계산 없음).

- **전제**: 각 idx의 **Part3(grad) 캐시가 있어야 함** (optim 완료). 없으면 그 idx는 skip.
- sweep: 진폭 6(0.5~3.0) × 주파수 5(20/125/143/167/200) = **30조합 × 4타깃**(STN_L/R, GP_L/R), 각 pre/during 600s
- 출력: `output_ppmi_pd/dbs_sweep/idx<i>/amp<a>_f<f>/{STN_L,STN_R,GP_L,GP_R}/fc_pre_during_diff.png`

그냥 아래 셀 실행 → idx 4,5,6 순차 sweep. (A10 Jupyter, forward-only라 Part3보다 가벼움)

```python
import os
os.environ.setdefault("XLA_PYTHON_CLIENT_PREALLOCATE", "false")
os.environ.setdefault("XLA_PYTHON_CLIENT_ALLOCATOR", "platform")

import sys
import dbs_from_cache as D

# ===== sweep 설정 =====
IDXS  = [4, 5, 6]
AMPS  = "0.5,1.0,1.5,2.0,2.5,3.0"     # 진폭 (모델 단위)
FREQS = "20,125,143,167,200"          # 주파수 Hz (dt=1ms 양자화)

for idx in IDXS:
    print(f"\n{'='*60}\n=== idx {idx} DBS sweep (30조합 × 4타깃) ===\n{'='*60}")
    sys.argv = [
        "dbs_from_cache",
        "--subject-idx", str(idx),
        "--dbs-amplitudes", AMPS,
        "--dbs-freqs", FREQS,
        "--dbs-outroot", f"output_ppmi_pd/dbs_sweep/idx{idx}",
    ]
    try:
        D.main()
    except SystemExit as e:
        print(f"  ⚠ idx {idx} skip (grad 캐시 없음 = Part3 미완?): {e}")
    except Exception as e:
        print(f"  ✗ idx {idx} 에러: {type(e).__name__}: {e}")

print("\n=== 전체 sweep 완료 → output_ppmi_pd/dbs_sweep/idx{4,5,6}/ ===")
```
